## **텍스트 분석**

### 1. 텍스트 정규화

```python
# 1. 텍스트 샘플(Text Sample) 로드 및 문장 토큰화(Sentence Tokenization)
#    자연어 처리 라이브러리인 NLTK를 활용하여 텍스트를 문장 단위로 분할하고 결과를 확인합니다.

import nltk
from nltk import sent_tokenize

# 토큰화에 필요한 자원(Resource) 다운로드
nltk.download('punkt')
nltk.download('punkt_tab')

# 분석할 텍스트 샘플 정의
text_sample = 'The Matrix is everywhere its all around us, here even in this room. \
              You can see it out your window or on your television. \
              You feel it when you go to work, or go to church or pay your taxes.'

# NLTK의 sent_tokenize를 이용해 문장 토큰화 수행
sentences = sent_tokenize(text=text_sample)

# 생성된 토큰 리스트의 자료형(Type) 및 문장 개수(Length) 출력하여 구조 확인
print(type(sentences), len(sentences))

# 분할된 문장 데이터 전체 출력
print(sentences)
```

```python
# 2. 문장 데이터 로드 및 단어 토큰화(Word Tokenization)
#    NLTK 라이브러리를 활용하여 단일 문장을 단어(Word) 단위로 분할하고 구조를 확인합니다.

from nltk import word_tokenize

# 분석할 문장 샘플 정의
sentence = "The Matrix is everywhere its all around us, here even in this room."

# NLTK의 word_tokenize를 이용해 단어 토큰화 수행
words = word_tokenize(sentence)

# 생성된 토큰 리스트의 자료형(Type) 및 단어 개수(Length) 출력하여 구조 확인
print(type(words), len(words))

# 분할된 단어 데이터 전체 출력
print(words)
```

```python
# 3. 문서 데이터 로드 및 종합 토큰화(Text Tokenization) 함수 구현
#    텍스트 샘플을 문장별로 먼저 분할한 후, 각 문장을 다시 단어 단위로 토큰화하는 함수를 정의하고 검증합니다.

from nltk import word_tokenize, sent_tokenize

# 여러 개의 문장으로 된 입력 데이터를 문장별로 단어 토큰화하는 사용자 정의 함수 생성
def tokenize_text(text):

    # 입력된 텍스트를 문장별로 분리(Sentence Tokenization)
    sentences = sent_tokenize(text)

    # 분리된 각 문장별로 단어 토큰화(Word Tokenization) 수행 후 리스트로 저장
    word_tokens = [word_tokenize(sentence) for sentence in sentences]
    return word_tokens

# 여러 문장에 대해 문장별 단어 토큰화 수행
word_tokens = tokenize_text(text_sample)

# 생성된 2차원 리스트의 자료형(Type) 및 문장 개수(Length) 출력하여 구조 확인
print(type(word_tokens), len(word_tokens))

# 문장별로 분할된 단어 토큰 데이터 전체 출력 (2차원 리스트 형태)
print(word_tokens)
```

```python
# 4. NLTK 불용어(Stopwords) 데이터 세트 다운로드
#    텍스트 전처리 과정에서 의미가 없는 단어를 제거하기 위해 NLTK가 제공하는 불용어 사전을 불러옵니다.

import nltk

# 영어, 한국어 등 다양한 언어의 불용어 사전 자원(Resource) 다운로드
nltk.download('stopwords')
```

```python
# 5. 영어 불용어(Stopwords) 데이터 세트 구조 및 샘플 확인
#    NLTK에 내장된 영어 불용어의 전체 개수를 확인하고, 상위 20개 샘플을 출력하여 데이터를 파악합니다.

# 등록된 영어 불용어(Stopwords)의 전체 개수(Length) 출력
print('영어 stop words 개수:', len(nltk.corpus.stopwords.words('english')))

# 영어 불용어 목록 중 상위 20개 단어(Row)를 출력하여 구조 확인
print(nltk.corpus.stopwords.words('english')[:20])
```

```python
# 6. 불용어(Stopwords) 제거를 통한 단어 필터링 전처리
#    앞서 생성한 문장별 단어 토큰 리스트에서 영어 불용어를 제거하고 실질적인 의미를 가진 단어만 추출합니다.

import nltk

# NLTK 영어 불용어(Stopwords) 목록 로드
stopwords = nltk.corpus.stopwords.words('english')
all_tokens = []

# 3개의 문장별로 분할된 word_tokens 리스트를 순회하며 불용어 제거 반복문 수행
for sentence in word_tokens:
    filtered_words = []

    # 개별 문장 내의 토큰화된 단어 리스트를 순회
    for word in sentence:
        # 대소문자 구분을 없애기 위해 소문자로 모두 변환(Lowercasing)
        word = word.lower()

        # 토큰화된 개별 단어가 영어 불용어 목록에 포함되지 않는 경우에만 추출
        if word not in stopwords:
            filtered_words.append(word)

    # 불용어가 제거된 단어 리스트를 최종 토큰 리스트에 추가
    all_tokens.append(filtered_words)

# 불용어가 정제된 문장별 단어 토큰 데이터 전체 출력하여 구조 확인
print(all_tokens)
```

```python
# 7. 어간 추출(Stemming) 모델을 활용한 단어 원형 추출
#    LancasterStemmer를 이용하여 다양한 형태로 변형된 단어들의 어간(Stem)을 추출하고 결과를 비교합니다.

from nltk.stem import LancasterStemmer

# LancasterStemmer 객체 생성
stemmer = LancasterStemmer()

# 'work' 관련 동사 변형들의 어간 추출 결과 출력
print(stemmer.stem('working'), stemmer.stem('works'), stemmer.stem('worked'))

# 'amuse' 관련 동사 변형들의 어간 추출 결과 출력
print(stemmer.stem('amusing'), stemmer.stem('amuses'), stemmer.stem('amused'))

# 'happy' 관련 형용사 비교급/최상급 변형들의 어간 추출 결과 출력
print(stemmer.stem('happier'), stemmer.stem('happiest'))

# 'fancy' 관련 형용사 비교급/최상급 변형들의 어간 추출 결과 출력하여 구조 확인
print(stemmer.stem('fancier'), stemmer.stem('fanciest'))
```

```python
# 8. 표제어 추출(Lemmatization) 모델을 활용한 단어 원형 복원
#    WordNetLemmatizer를 활용하여 단어의 품사(POS) 정보를 기반으로 정확한 표제어(Lemma)를 추출합니다.

import nltk
from nltk.stem import WordNetLemmatizer

# 표제어 추출에 필요한 WordNet 데이터 세트(Resource) 다운로드
nltk.download('wordnet')

# WordNetLemmatizer 객체 생성
lemma = WordNetLemmatizer()

# 동사('v') 품사 정보를 지정하여 'amuse' 관련 단어들의 표제어 추출 결과 출력
print(lemma.lemmatize('amusing', 'v'), lemma.lemmatize('amuses', 'v'), lemma.lemmatize('amused', 'v'))

# 형용사('a') 품사 정보를 지정하여 'happy' 관련 단어들의 표제어 추출 결과 출력
print(lemma.lemmatize('happier', 'a'), lemma.lemmatize('happiest', 'a'))

# 형용사('a') 품사 정보를 지정하여 'fancy' 관련 단어들의 표제어 추출 결과 출력하여 구조 확인
print(lemma.lemmatize('fancier', 'a'), lemma.lemmatize('fanciest', 'a'))
```

### 2. Bag of Words - BOW

```python
# 1. 넘파이(NumPy) 기반 밀집 행렬(Dense Matrix) 생성 및 데이터 정의
#    희소 행렬 변환을 수행하기 전, 다차원 배열을 활용하여 기본 밀집 행렬의 구조를 정의합니다.

import numpy as np

# 2행 3열 구조의 2차원 밀집 행렬 데이터 생성
dense = np.array([[3, 0, 1], [0, 2, 0]])
```

```python
# 2. COO(Coordinate) 형식을 활용한 희소 행렬(Sparse Matrix) 생성
#    SciPy 라이브러리의 coo_matrix를 사용하여 0이 아닌 데이터와 해당 위치(행, 열) 정보를 기반으로 희소 행렬을 구조화합니다.

from scipy import sparse

# 0이 아닌 실제 데이터 값(Data) 배열 생성
data = np.array([3, 1, 2])

# 데이터가 위치한 행 인덱스(Row Position)와 열 인덱스(Column Position) 배열을 각각 생성
row_pos = np.array([0, 0, 1])
col_pos = np.array([0, 2, 1])

# SciPy의 coo_matrix를 이용하여 COO 형식의 희소 행렬 변환 수행
sparse_coo = sparse.coo_matrix((data, (row_pos, col_pos)))
```

```python
# 3. 희소 행렬(Sparse Matrix)의 밀집 행렬(Dense Matrix) 변환 및 구조 확인
#    COO 형식으로 압축된 희소 행렬을 toarray() 메서드를 통해 다시 원래의 넘파이 배열로 복원합니다.

# COO 형식의 희소 행렬을 일반 다차원 넘파이 배열(Array) 형태로 변환하여 출력
sparse_coo.toarray()
```

```python
# 4. COO 및 CSR 형식을 활용한 희소 행렬(Sparse Matrix) 변환 및 검증
#    6x6 밀집 행렬을 COO 형식과 고유 인덱스 배열을 활용한 CSR 형식으로 각각 변환하고, 원래 배열로 복원하여 정확성을 검증합니다.

import numpy as np
from scipy import sparse

# 6행 6열 구조의 2차원 밀집 행렬(Dense Matrix) 데이터 정의
dense2 = np.array([[0, 0, 1, 0, 0, 5],
                   [1, 4, 0, 3, 2, 5],
                   [0, 6, 0, 3, 0, 0],
                   [2, 0, 0, 0, 0, 0],
                   [0, 0, 0, 7, 0, 8],
                   [1, 0, 0, 0, 0, 0]])

# 행렬 내에서 0이 아닌 실제 데이터 값(Data) 배열 생성
data2 = np.array([1, 5, 1, 4, 3, 2, 5, 6, 3, 2, 7, 8, 1])

# 데이터가 위치한 행 인덱스(Row Position)와 열 인덱스(Column Position) 배열을 각각 생성
row_pos = np.array([0, 0, 1, 1, 1, 1, 1, 2, 2, 3, 4, 4, 5])
coI_pos = np.array([2, 5, 0, 1, 3, 4, 5, 1, 3, 0, 3, 5, 0])

# SciPy의 coo_matrix를 이용하여 COO 형식의 희소 행렬 변환 수행
sparse_coo = sparse.coo_matrix((data2, (row_pos, coI_pos)))

# 행 위치 배열 내 고유한 값의 시작 위치 인덱스(Row Position Index) 배열 생성 (CSR 압축용)
row_pos_ind = np.array([0, 2, 7, 9, 10, 12, 13])

# SciPy의 csr_matrix를 이용하여 CSR 형식의 희소 행렬 변환 수행
sparse_csr = sparse.csr_matrix((data2, coI_pos, row_pos_ind))

# COO 형식으로 변환된 희소 행렬을 다시 밀집 행렬(Dense Matrix)로 복원하여 결과 출력
print('COO 변환된 데이터가 제대로 되었는지 다시 Dense로 출력 확인')
print(sparse_coo.toarray())

# CSR 형식으로 변환된 희소 행렬을 다시 밀집 행렬(Dense Matrix)로 복원하여 결과 출력 및 구조 확인
print('CSR 변환된 데이터가 제대로 되었는지 다시 Dense로 출력 확인')
print(sparse_csr.toarray())
```

```python
# 5. SciPy 함수 기반 COO 및 CSR 희소 행렬(Sparse Matrix) 자동 변환
#    위치 배열을 직접 생성하지 않고, SciPy 내장 함수를 활용하여 밀집 행렬을 COO와 CSR 형식으로 효율적으로 변환합니다.

import numpy as np
from scipy import sparse

# 6행 6열 구조의 2차원 밀집 행렬(Dense Matrix) 데이터 정의
dense3 = np.array([[0, 0, 1, 0, 0, 5],
                   [1, 4, 0, 3, 2, 5],
                   [0, 6, 0, 3, 0, 0],
                   [2, 0, 0, 0, 0, 0],
                   [0, 0, 0, 7, 0, 8],
                   [1, 0, 0, 0, 0, 0]])

# SciPy의 coo_matrix 인터페이스를 통해 밀집 행렬을 COO 형식 희소 행렬로 자동 변환
coo = sparse.coo_matrix(dense3)

# SciPy의 csr_matrix 인터페이스를 통해 밀집 행렬을 CSR 형식 희소 행렬로 자동 변환하여 구조화
csr = sparse.csr_matrix(dense3)
```

### 3. 감성 분석

```python
# 1. 감성 분석(Sentiment Analysis) 데이터 세트 로드 및 데이터프레임 변환
#    지도학습 기반의 감성 분석 모델 학습을 위해 탭(\t)으로 구분된 영화 리뷰 데이터를 불러오고 구조를 확인합니다.

import pandas as pd

# 영화 리뷰 데이터 세트 가져오기 (TSV 파일 로드)
review_df = pd.read_csv('/content/labeledTrainData.tsv', header=0, sep='\t', quoting=3)

# 생성된 데이터프레임(DataFrame)의 상위 3개 행(Row) 출력하여 데이터 구조 확인
review_df.head(3)
```

```python
# 2. 개별 리뷰 텍스트 데이터 추출 및 원문 확인
#    데이터프레임의 'review' 컬럼에서 첫 번째 인덱스의 데이터를 선택하여 분석 대상 텍스트의 형태를 파악합니다.

# 첫 번째 행(Row)의 영화 리뷰 텍스트(Text) 데이터를 선택하여 콘솔에 출력 및 확인
print(review_df['review'][0])
```

```python
# 3. 정규 표현식(Regular Expression)을 활용한 텍스트 데이터 정제 및 전처리
#    텍스트 내의 불필요한 HTML 태그와 영문이 아닌 특수문자, 숫자 등을 제거하여 감성 분석에 적합한 형태로 정제합니다.

import re

# <br /> HTML 태그를 replace 함수를 이용하여 공백(Space)으로 변환
review_df['review'] = review_df['review'].str.replace('<br />', ' ')

# 파이썬의 정규 표현식 모듈인 re를 이용해 영어 문자열이 아닌 모든 문자를 공백으로 변환 (정제 과정 수행)
review_df['review'] = review_df['review'].apply(lambda x : re.sub("[^a-zA-Z]", " ", x))
```

```python
# 4. 데이터 세트 분할 및 학습/테스트 데이터(Train/Test Dataset) 생성
#    학습용 피처/클래스 데이터와 검증용 피처/클래스 데이터를 7:3 비율로 분할하고 최종 데이터 구조를 확인합니다.

from sklearn.model_selection import train_test_split

# 타깃 레이블인 감성(Sentiment) 클래스 데이터프레임 추출
class_df = review_df['sentiment']

# 원본 데이터에서 불필요한 고유 ID와 타깃 식별자를 삭제하여 피처(Feature) 데이터프레임 생성
feature_df = review_df.drop(['id', 'sentiment'], axis=1, inplace=False)

# train_test_split을 이용하여 학습 데이터 세트와 테스트 데이터 세트로 분할 (테스트 비율 30%)
X_train, X_test, y_train, y_test = train_test_split(feature_df, class_df, test_size=0.3, random_state=156)

# 분할된 학습 피처 데이터와 테스트 피처 데이터의 행렬 구조(Shape) 확인
X_train.shape, X_test.shape
```

```python
# 5. 파이프라인(Pipeline)을 활용한 카운트 기반 피처 벡터화 및 로지스틱 회귀 학습/평가
#    텍스트 데이터의 Count 벡터화와 로지스틱 회귀(Logistic Regression) 모델의 학습, 예측, 성능 평가를 단일 파이프라인으로 결합하여 수행합니다.

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, roc_auc_score

# CountVectorizer(피처 벡터화)와 LogisticRegression(분류 모델)을 결합한 파이프라인 객체 생성
# 스톱 워드는 English, n-gram은 (1, 2), 분류기의 규제 계수 C는 10으로 설정
pipeline = Pipeline([
    ('cnt_vect', CountVectorizer(stop_words='english', ngram_range=(1, 2))),
    ('lr_clf', LogisticRegression(solver='liblinear', C=10))
])

# 학습 데이터의 리뷰 텍스트를 입력하여 파이프라인 전체 프로세스(벡터화 및 모델) 학습(Fit) 수행
pipeline.fit(X_train['review'], y_train)

# 테스트 데이터를 이용해 클래스 예측(Predict) 및 긍정 확률(Predict Proba) 추출
pred = pipeline.predict(X_test['review'])
pred_probs = pipeline.predict_proba(X_test['review'])[:, 1]

# 모델의 최종 성능 지표인 예측 정확도(Accuracy) 및 ROC-AUC 수치를 출력하여 구조 확인
print('예측 정확도는 {0:.4f}, ROC-AUC는 {1:.4f}'.format(accuracy_score(y_test, pred), roc_auc_score(y_test, pred_probs)))
```

```python
# 6. NLTK 전체 코퍼스(Corpus) 및 데이터 세트 통합 다운로드
#    자연어 처리에 필요한 NLTK 라이브러리의 모든 패키지, 사전 데이터, 언어 모델 자원(Resource)을 일괄 다운로드합니다.

import nltk

# NLTK가 제공하는 전체 데이터 세트('all') 일괄 다운로드 수행
nltk.download('all')
```

```python
# 7. WordNet 기반 단어의 시셋(Synset) 추출 및 의미 집합 구조 확인
#    NLTK의 WordNet을 활용하여 특정 단어가 내포하는 다의어/동음이의어 사전 구조를 불러오고 개수를 파악합니다.

from nltk.corpus import wordnet as wn

# 분석할 기준 단어(Term) 정의
term = 'present'

# 'present'라는 단어를 기반으로 WordNet에 등록된 시셋(Synsets) 리스트 생성
synsets = wn.synsets(term)

# 반환된 시셋 객체의 자료형(Type) 출력
print('synsets() 반환 type :', type(synsets))

# 해당 단어가 가지고 있는 고유한 의미 집합의 개수(Length) 출력
print('synsets() 반환 값 갯수:', len(synsets))

# 단어가 가진 모든 어휘적 의미(Synset) 목록 전체 출력하여 구조 확인
print('synsets() 반환 값 :', synsets)
```

```python
# 8. 시셋(Synset) 객체별 어휘 세부 정보 및 의미론적 구조 분석
#    추출된 단어 의미 집합 리스트를 순회하며 각 시셋의 이름, 품사 분류(POS), 개념 정의(Definition), 표제어 목록을 출력합니다.

# 단어 'present'로 생성된 모든 시셋(Synsets) 리스트를 순회하는 반복문
for synset in synsets:
    # 개별 시셋의 고유 식별 명칭(Name) 출력
    print('##### Synset name : ', synset.name(), '#####')

    # 해당 의미가 속한 문맥상의 품사 및 분류 그룹(Lexicographer Name) 출력
    print('POS :', synset.lexname())

    # 해당 어휘 의미에 대한 사전적 주석 및 개념 정의(Definition) 출력
    print('Definition:', synset.definition())

    # 동일한 의미를 공유하는 유의어 및 표제어(Lemmas) 리스트를 출력하여 구조 확인
    print('Lemmas:', synset.lemma_names())
```

```python
# 9. WordNet 시셋(Synset) 간의 경로 유사도(Path Similarity) 측정 및 데이터프레임 매핑
#    개별 단어의 시셋 객체를 생성하고, 상호 간의 의미론적 경로 유사도를 계산하여 행렬 구조의 데이터프레임으로 시각화합니다.

import pandas as pd
from nltk.corpus import wordnet as wn

# 비교 분석할 주요 단어별 시셋(Synset) 객체 생성
tree = wn.synset('tree.n.01')
lion = wn.synset('lion.n.01')
tiger = wn.synset('tiger.n.02')
cat = wn.synset('cat.n.01')
dog = wn.synset('dog.n.01')

# 시셋 객체 리스트 및 유사도 측정값 저장용 리스트 정의
entities = [tree, lion, tiger, cat, dog]
similarities = []

# 각 시셋의 고유 식별 명칭에서 순수 단어명(Entity Name)만 추출하여 리스트 구성
entity_names = [entity.name().split('.')[0] for entity in entities]

# 단어별 시셋 리스트를 순회하며 다른 단어의 시셋과의 경로 유사도(Path Similarity) 반복 측정
for entity in entities:
    # 소수점 둘째 자리까지 반올림하여 상호 간의 유사도 행렬 계산
    similarity = [round(entity.path_similarity(compared_entity), 2)
                  for compared_entity in entities]
    similarities.append(similarity)

# 계산된 단어별 유사도 리스트 데이터를 판다스 데이터프레임(DataFrame) 구조로 변환 및 컬럼/인덱스 명명
similarity_df = pd.DataFrame(similarities, columns=entity_names, index=entity_names)

# 생성된 유사도 행렬 데이터프레임 전체를 출력하여 구조 확인
similarity_df
```

```python
# 10. SentiWordNet 기반 감성 시셋(SentiSynset) 추출 및 어휘 감성 구조 확인
#     단어가 가진 문맥별 감성 지수(긍정/부정/객관성)를 분석하기 위해 특정 단어의 감성 시셋 리스트를 생성하고 구조를 파악합니다.

import nltk
from nltk.corpus import sentiwordnet as swn

# 'slow'라는 단어를 기반으로 SentiWordNet에 등록된 감성 시셋(SentiSynsets) 목록을 리스트로 변환
senti_synsets = list(swn.senti_synsets('slow'))

# 반환된 감성 시셋 리스트의 자료형(Type) 출력
print('senti_synsets() 반환 type :', type(senti_synsets))

# 해당 단어가 가지고 있는 고유한 감성 의미 집합의 개수(Length) 출력
print('senti_synsets() 반환 값 갯수:', len(senti_synsets))

# 어휘별 감성 주석이 포함된 감성 시셋 데이터 전체를 출력하여 구조 확인
print('senti_synsets() 반환 값 :', senti_synsets)
```

```python
# 11. SentiSynset 객체의 감성 지수(Sentiment Score) 추출 및 수치 분석
#     특정 어휘 시셋의 긍정 지수(Positive Score), 부정 지수(Negative Score), 객관성 지수(Objective Score)를 조회하여 감정 상태를 정량화합니다.

import nltk
from nltk.corpus import sentiwordnet as swn

# 명사('n') 품사 정보를 가진 'father'의 첫 번째 감성 시셋 객체 생성
father = swn.senti_synset('father.n.01')

# 'father' 시셋의 감성 지수(긍정, 부정, 객관성)를 각각 추출하여 출력
print('father 긍정감성 지수:', father.pos_score())
print('father 부정감성 지수:', father.neg_score())
print('father 객관성 지수:', father.obj_score())

print('\n')

# 형용사('a') 품사 정보를 가진 'fabulous'의 첫 번째 감성 시셋 객체 생성
fabulous = swn.senti_synset('fabulous.a.01')

# 'fabulous' 시셋의 감성 지수(긍정, 부정)를 각각 출력하여 구조 확인
print('fabulous 긍정감성 지수:', fabulous.pos_score())
print('fabulous 부정감성 지수:', fabulous.neg_score())
```

```python
# 12. Penn Treebank 품사 태그의 WordNet 품사 식별자 변환 함수 구현
#     NLTK의 pos_tag()가 반환하는 Penn Treebank 형식의 접두사를 판별하여 WordNet 언어 모델 호환 품사(POS) 태그로 매핑합니다.

from nltk.corpus import wordnet as wn

# Penn Treebank 품사 Tag를 기반으로 WordNet 기반의 품사 Tag로 변환하는 함수 생성
def penn_to_wn(tag):

    # 접두사가 'J'로 시작하는 경우 형용사(Adjective) 식별자 반환
    if tag.startswith('J'):
        return wn.ADJ

    # 접두사가 'N'으로 시작하는 경우 명사(Noun) 식별자 반환
    elif tag.startswith('N'):
        return wn.NOUN

    # 접두사가 'R'로 시작하는 경우 부사(Adverb) 식별자 반환
    elif tag.startswith('R'):
        return wn.ADV

    # 접두사가 'V'로 시작하는 경우 동사(Verb) 식별자 반환 구조 정의
    elif tag.startswith('V'):
        return wn.VERB
```

```python
# 13. SentiWordNet 기반 문서 감성 극성(Polarity) 예측 함수 구현
#     문장을 토큰화하고 품사 태깅과 어간 추출을 거쳐 단어별 긍정/부정 지수를 합산한 뒤, 최종 감성 스코어를 판별하는 함수를 정의합니다.

from nltk.stem import WordNetLemmatizer
from nltk.corpus import sentiwordnet as swn
from nltk.corpus import wordnet as wn
from nltk import sent_tokenize, word_tokenize, pos_tag

# 텍스트 데이터를 입력받아 감성 극성을 반환하는 사용자 정의 함수 생성
def swn_polarity(text):
    # 감성 지수 및 연산에 사용된 토큰 개수(Count) 초기화
    sentiment = 0.0
    tokens_count = 0

    lemmatizer = WordNetLemmatizer()
    raw_sentences = sent_tokenize(text)

    # 분해된 문장별로 단어 토큰화 후 품사 태깅, SentiSynset 생성 및 감성 지수 합산 루프 수행
    for raw_sentence in raw_sentences:
        # NLTK 기반의 품사 태깅(POS Tagging) 수행
        tagged_sentence = pos_tag(word_tokenize(raw_sentence))
        for word, tag in tagged_sentence:

            # 이전에 정의한 penn_to_wn 함수를 활용해 WordNet 기반 품사 태그로 변환
            wn_tag = penn_to_wn(tag)
            if wn_tag not in (wn.NOUN, wn.ADJ, wn.ADV):
                continue

            # 변환된 품사 정보('pos')를 지정하여 단어 어근 추출(Lemmatization) 수행
            lemma = lemmatizer.lemmatize(word, pos=wn_tag)
            if not lemma:
                continue

            # 어근을 추출한 단어와 WordNet 기반 품사 태깅을 입력해 시셋(Synset) 객체 생성
            synsets = wn.synsets(lemma, pos=wn_tag)
            if not synsets:
                continue

            # SentiWordNet의 감성 단어 분석을 위해 매칭되는 첫 번째 감성 시셋(SentiSynset) 추출
            synset = synsets[0]
            swn_synset = swn.senti_synset(synset.name())

            # 모든 단어에 대해 긍정 감성 지수는 더하고(+) 부정 감성 지수는 빼서(-) 감성 총합 계산
            sentiment += (swn_synset.pos_score() - swn_synset.neg_score())
            tokens_count += 1

    # 분석에 유효한 감성 토큰이 전혀 없을 경우 0 반환
    if not tokens_count:
        return 0

    # 총 스코어(Score)가 0 이상일 경우 긍정(Positive)인 1, 그렇지 않을 경우 부정(Negative)인 0 반환 구조 정의
    if sentiment >= 0:
        return 1

    return 0
```

```python
# 14. 전체 데이터 세트 대상 감성 극성 예측 수행 및 결과 배열 추출
#     정의된 감성 분석 함수를 데이터프레임의 모든 리뷰에 적용(Apply)하여 예측 평점(Prediction)을 도출하고 최종 분류용 타깃/예측 배열을 생성합니다.

# 데이터프레임의 'review' 컬럼 전체 텍스트에 감성 분석 람다(Lambda) 함수를 적용하여 'preds' 컬럼에 결과 저장
review_df['preds'] = review_df['review'].apply(lambda x : swn_polarity(x))

# 성능 평가 비교를 위해 실제 정답 레이블인 감성(Sentiment) 데이터를 넘파이 배열(Values) 형태로 추출
y_target = review_df['sentiment'].values

# 모델을 통해 예측된 감성 극성 결과(Predictions) 데이터를 넘파이 배열 형태로 추출하여 구조 확인
preds = review_df['preds'].values
```

```python
# 16. VADER 감성 분석기(Sentiment Intensity Analyzer)를 활용한 리뷰 감성 지수 산출
#     소셜 미디어 및 텍스트 문맥 분석에 특화된 VADER 알고리즘을 사용하여 단어의 극성 점수를 구하고 다각적 감성 구조를 측정합니다.

from nltk.sentiment.vader import SentimentIntensityAnalyzer

# VADER 감성 분석기(SentimentIntensityAnalyzer) 객체 생성 및 인스턴스화
senti_analyzer = SentimentIntensityAnalyzer()

# 첫 번째 영화 리뷰 텍스트 원문을 입력하여 감성 극성 점수(Polarity Scores) 계산
senti_scores = senti_analyzer.polarity_scores(review_df['review'][0])

# 계산된 세부 감성 지수(긍정/부정/중립/복합 스코어) 딕셔너리 결과를 출력하여 구조 확인
print(senti_scores)
```

```python
# 17. VADER 기반 문서 감성 극성(Polarity) 분류 함수 구현 및 최종 성능 검증
#     임계값(Threshold) 설정을 통해 복합 감성 지수인 compound 스코어를 0(부정) 또는 1(긍정)로 이진 분류하고 최종 평가지표를 연산합니다.

from sklearn.metrics import accuracy_score, confusion_matrix, precision_score
from sklearn.metrics import recall_score, f1_score, roc_auc_score
import numpy as np

# 리뷰 텍스트와 임계값을 입력받아 VADER 감성 극성을 반환하는 사용자 정의 함수 생성
def vader_polarity(review, threshold=0.1):
    analyzer = SentimentIntensityAnalyzer()
    scores = analyzer.polarity_scores(review)

    # 복합 감성 지수인 compound 값에 기반해 threshold 입력값보다 크거나 같으면 1, 그렇지 않으면 0 반환
    agg_score = scores['compound']
    final_sentiment = 1 if agg_score >= threshold else 0
    return final_sentiment

# apply lambda 식을 이용하여 레코드별로 vader_polarity()를 수행하고 결과를 'vader_preds' 컬럼에 저장
review_df['vader_preds'] = review_df['review'].apply(lambda x : vader_polarity(x, 0.1))

# 성능 평가 비교를 위해 실제 정답 레이블과 VADER 예측값 배열을 각각 추출
y_target = review_df['sentiment'].values
vader_preds = review_df['vader_preds'].values

# VADER 모델의 오차 행렬(Confusion Matrix) 구조 출력
print(confusion_matrix(y_target, vader_preds))

# 최종 분류 모델 핵심 성능 지표(정확도, 정밀도, 재현율)를 반올림하여 출력 및 확인
print("정확도:", np.round(accuracy_score(y_target, vader_preds), 4))
print("정밀도:", np.round(precision_score(y_target, vader_preds), 4))
print("재현율:", np.round(recall_score(y_target, vader_preds), 4))
```
